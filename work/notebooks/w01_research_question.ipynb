## 2. The question: decision, action, cost of a wrong call

The Core Research Question: Which existing content pages are showing clear early signals of traffic decay, and how should we prioritize them for content refresh/editorial updates?

Unit of Analysis: A single page/URL over a 90-day performance window.

The Business Decision: Deciding which 50 to 100 pages out of thousands should receive editorial resources and budget for content updates this month.

The Action: An editor or SEO strategist updates the content (e.g., refreshing outdated facts, improving search intent match, adding internal links).

Cost of a Wrong Call:

False Positive (recommending a healthy page for refresh): Wastes expensive editorial time and budget on pages that don't need work, with a risk of breaking something that is already ranking well.

False Negative (missing a decaying page): The page continues to lose rankings and traffic to competitors, leading to permanent organic revenue loss.

Why ML helps: Simple static rules (e.g., "update pages older than 1 year") ignore complex interactions between impressions, position drops, CTR collapse, and intent shifts. ML can weigh multiple subtle decay signals simultaneously.
